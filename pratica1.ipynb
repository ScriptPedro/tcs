```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import root_mean_squared_error, mean_squared_error, r2_score
from sklearn.model_selection import KFold

```

```python
dataset = load_diabetes()

x = dataset.data
y = dataset.target
```

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)
model = LinearRegression()

for i, (train_index, test_index) in enumerate(kf.split(x)):
    model.fit(x[train_index], y[train_index])
    
    result = model.predict(x[test_index])
    
    mse = mean_squared_error(y[test_index], result)
    r2 = r2_score(y[test_index], result)
    rmse = root_mean_squared_error(y[test_index], result)
    
   # print(f"Fold {i + 1}:")
   # print(f"Previsões: {result}")
   # print(f"Valores reais: {y[test_index]}")
    print(f"MSE: {mse:.4f}")
    print(f"R²: {r2:.4f}")
    print(f"RMSE: {rmse:.4f}")
    print("-" * 30)
```
